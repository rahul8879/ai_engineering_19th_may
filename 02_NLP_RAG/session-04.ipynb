```python
# pip install sentence_transformers
from sentence_transformers import SentenceTransformer
model = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
sentences = [
    "Gold loan foreclosure charge is 2 percent within 6 months.",
    "Closing a gold loan early may attract a 2 percent penalty in the first 6 months.",
    "Personal loan EMI must be paid every month.",
]

vector = model.encode(sentences)
```

```python
chunk_texts = [
    "Gold loan foreclosure attracts 2 percent penalty if closed within 6 months.",
    "Customers must pay personal loan EMI on the due date every month.",
    "A low CIBIL score may reduce the chance of loan approval.",
    "Missing EMI payments can lead to penalty charges and collection follow-up.",
]


chunk_vectors = model.encode(chunk_texts)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
def search_documents(query, top_k=2):
    # convert query into one embedding vectors
    query_vector = model.encode([query])
    # compare query with all chunks embeddings ?
    similarities = cosine_similarity(query_vector, chunk_vectors)[0]
    # get top matching indices
    top_k_indices = similarities.argsort()[-top_k:][::-1]
    # return [chunk_texts[i] for i in top_k_indices]
    result = []
    for i in top_k_indices:
        result.append((chunk_texts[i], similarities[i]))
    return result
```

```python
query = "Hi"
search_documents(query)
```

```python
search_documents(query)
```

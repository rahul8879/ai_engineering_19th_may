```python
import numpy as np
import matplotlib.pyplot as plt


not_paid = np.array([[2,4],
                     [3,3],
                     [4,5],
                     [4,4]])


paid = np.array([[7,8],
                 [8,8],
                 [7,8],
                 [7,6]])

```

```python
# parameters
a,b,c = -0.6, -0.4, 5.0
```

```python
fig,ax = plt.subplots(figsize=(8,6))
ax.scatter(not_paid[:,0], not_paid[:,1], color='red', label='Not Paid')
ax.scatter(paid[:,0], paid[:,1], color='green', label='Paid')
x_vals = np.linspace(0, 10, 100)
y_vals = (-a/b)*x_vals - c/b
ax.plot(x_vals, y_vals, color='blue', label='Decision Boundary')
ax.set_xlabel('Feature 1')
ax.set_ylabel('Feature 2')
ax.set_title('Paid vs Not Paid')
ax.legend()
```

```python
def predict(features):
    x1, x2 = features
    if (a * x1 + b * x2 + c) > 0:
        return 1  # Paid
    else:
        return 0  # Not Paid
```

```python
predict([8, 8])
```

```python
# you can plot the trends
import matplotlib.pyplot as plt
data= sales_data[sales_data['sku'] == 'SKU060']
# lets sort the date
data = data.sort_values('date')

# normalize
data['qty'] = (data['qty'] - data['qty'].mean()) / data['qty'].std()

plt.figure(figsize=(12, 6))
plt.plot(data['date'], data['qty'], marker='o')
plt.title('Sales Trend for SKU060')
plt.xlabel('Date')
plt.ylabel('Sales')
plt.xticks(rotation=45)
plt.grid()
plt.show()
```
